```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("pakistan_development_processed.csv")

df["Year"] = pd.to_numeric(df["Year"])

df.head()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Year"],
    df["GDP Growth"],
    linewidth=2
)

plt.axhline(0, linestyle="--", linewidth=1)

plt.title("Pakistan GDP Growth Over Time")
plt.xlabel("Year")
plt.ylabel("GDP Growth (%)")

plt.tight_layout()

plt.savefig("gdp_growth_pakistan.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Year"],
    df["Population"],
    linewidth=2
)

plt.title("Population Growth in Pakistan")
plt.xlabel("Year")
plt.ylabel("Population")

plt.tight_layout()
plt.savefig("population_pakistan.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Year"],
    df["Life Expectancy"],
    linewidth=2
)

plt.title("Life Expectancy in Pakistan Over Time")
plt.xlabel("Year")
plt.ylabel("Life Expectancy (Years)")
plt.savefig("life_expectancy_pakistan.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Year"],
    df["GDP per Capita"],
    linewidth=2
)

plt.title("GDP per Capita in Pakistan Over Time")
plt.xlabel("Year")
plt.ylabel("GDP per Capita (Current US$)")
plt.savefig("gdp_per_capita_pakistan.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Year"],
    df["Urban Population"],
    linewidth=2
)

plt.title("Urban Population in Pakistan Over Time")
plt.xlabel("Year")
plt.ylabel("Urban Population (% of Total)")
plt.savefig("urban_population_pakistan.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Year"],
    df["Access to Electricity"],
    marker="o",
    linewidth=2
)

plt.title("Access to Electricity in Pakistan")
plt.xlabel("Year")
plt.ylabel("Access to Electricity (%)")
plt.savefig("electricity_access_pakistan.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Year"],
    df["Renewable Energy"],
    marker="o",
    linewidth=2
)

plt.title("Renewable Energy Consumption in Pakistan")
plt.xlabel("Year")
plt.ylabel("Renewable Energy (% of Total Energy)")
plt.savefig("renewable_energy_pakistan.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

correlation
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Between Pakistan Development Indicators")

plt.tight_layout()

plt.savefig(
    "development_correlation.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
