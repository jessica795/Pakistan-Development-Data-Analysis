```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("pakistan_development_processed.csv")
df["Year"] = pd.to_numeric(df["Year"])

pd.set_option("display.max_columns", None)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].plot(
    df["Year"],
    df["GDP per Capita"],
    linewidth=2
)
axes[0, 0].set_title("GDP per Capita")
axes[0, 0].set_xlabel("Year")
axes[0, 0].set_ylabel("Current US$")

axes[0, 1].plot(
    df["Year"],
    df["Life Expectancy"],
    linewidth=2
)
axes[0, 1].set_title("Life Expectancy")
axes[0, 1].set_xlabel("Year")
axes[0, 1].set_ylabel("Years")

axes[1, 0].plot(
    df["Year"],
    df["Urban Population"],
    linewidth=2
)
axes[1, 0].set_title("Urban Population")
axes[1, 0].set_xlabel("Year")
axes[1, 0].set_ylabel("% of Total Population")

axes[1, 1].plot(
    df["Year"],
    df["Access to Electricity"],
    linewidth=2
)
axes[1, 1].set_title("Access to Electricity")
axes[1, 1].set_xlabel("Year")
axes[1, 1].set_ylabel("% of Population")

plt.suptitle(
    "Selected Development Indicators in Pakistan",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
summary = pd.DataFrame({
    "Indicator": [
        "GDP per Capita",
        "Life Expectancy",
        "Urban Population",
        "Access to Electricity",
        "Renewable Energy"
    ],
    "First Available Value": [
        df["GDP per Capita"].dropna().iloc[0],
        df["Life Expectancy"].dropna().iloc[0],
        df["Urban Population"].dropna().iloc[0],
        df["Access to Electricity"].dropna().iloc[0],
        df["Renewable Energy"].dropna().iloc[0]
    ],
    "Latest Available Value": [
        df["GDP per Capita"].dropna().iloc[-1],
        df["Life Expectancy"].dropna().iloc[-1],
        df["Urban Population"].dropna().iloc[-1],
        df["Access to Electricity"].dropna().iloc[-1],
        df["Renewable Energy"].dropna().iloc[-1]
    ]
})

summary
```
