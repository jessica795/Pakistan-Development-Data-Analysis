```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

df = pd.read_csv("pakistan_development_raw.csv")

df.head()
```

```python
print("Dataset shape:", df.shape)

df.info()
```

```python
missing_values = df.isna().sum()

missing_values
```

```python
missing_percentage = (
    df.isna().mean() * 100
).sort_values(ascending=False)

missing_percentage
```

```python
duplicate_rows = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_rows)
```

```python
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

print(df["Year"].min())
print(df["Year"].max())
```

```python
df["Year"].isna().sum()
```

```python
descriptive_statistics = df.describe().T

descriptive_statistics
```

```python
descriptive_statistics.to_csv("descriptive_statistics.csv", index=True)

print("Descriptive statistics saved.")
```

```python
# Save the cleaned dataset
df.to_csv("pakistan_development_processed.csv", index=False)

print("Processed dataset saved successfully.")
```
